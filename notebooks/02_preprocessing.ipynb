```python
import sys
sys.path.append("../")

import pandas as pd
from src.preprocess import (
    load_hard_data, binarize_labels, keep_required_columns,
    apply_preprocessing, preprocess_for_tfidf, preprocess_for_arabert
)
```

```python
df_bal_raw   = load_hard_data("../data/balanced/balanced-reviews.txt")
df_unbal_raw = load_hard_data("../data/unbalanced/unbalanced-reviews.txt")

df_bal   = keep_required_columns(binarize_labels(df_bal_raw))
df_unbal = keep_required_columns(binarize_labels(df_unbal_raw))

print("Balanced:  ", df_bal.shape)
print("Unbalanced:", df_unbal.shape)
```

```python
sample_reviews = df_bal["review"].sample(5, random_state=42).tolist()

print("BEFORE → AFTER (TF-IDF preprocessing)\n")
for rev in sample_reviews:
    print(f"  BEFORE: {rev[:150]}")
    print(f"  AFTER:  {preprocess_for_tfidf(rev)[:150]}")
    print()

print("\nBEFORE → AFTER (AraBERT preprocessing)\n")
for rev in sample_reviews:
    print(f"  BEFORE: {rev[:150]}")
    print(f"  AFTER:  {preprocess_for_arabert(rev)[:150]}")
    print()
```

```python
print("Applying TF-IDF preprocessing to balanced dataset...")
df_bal_tfidf = apply_preprocessing(df_bal, mode="tfidf")

print("Applying TF-IDF preprocessing to unbalanced dataset...")
df_unbal_tfidf = apply_preprocessing(df_unbal, mode="tfidf")

print(f"Balanced after preprocessing:   {df_bal_tfidf.shape}")
print(f"Unbalanced after preprocessing: {df_unbal_tfidf.shape}")

# ── Cell 5: Apply AraBERT Preprocessing ──────────────────────
print("Applying AraBERT preprocessing to balanced dataset...")
df_bal_arabert = apply_preprocessing(df_bal, mode="arabert")

print("Applying AraBERT preprocessing to unbalanced dataset...")
df_unbal_arabert = apply_preprocessing(df_unbal, mode="arabert")
```

```python
df_bal_tfidf.to_csv("../data/balanced/preprocessed_tfidf.csv", index=False, encoding="utf-8-sig")
df_unbal_tfidf.to_csv("../data/unbalanced/preprocessed_tfidf.csv", index=False, encoding="utf-8-sig")

df_bal_arabert.to_csv("../data/balanced/preprocessed_arabert.csv", index=False, encoding="utf-8-sig")
df_unbal_arabert.to_csv("../data/unbalanced/preprocessed_arabert.csv", index=False, encoding="utf-8-sig")

print("All preprocessed files saved.")
```

```python
print("\nLabel distribution after preprocessing:")
print("Balanced:  ", df_bal_tfidf["label"].value_counts().to_dict())
print("Unbalanced:", df_unbal_tfidf["label"].value_counts().to_dict())
```
